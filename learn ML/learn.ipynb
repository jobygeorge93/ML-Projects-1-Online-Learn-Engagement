## Introduction

Online learning has become an essential part of modern education, but student engagement varies widely. Understanding how learners interact with online courses helps educators improve teaching quality, personalize learning, and reduce dropout rates.
In this project, a machine learning model is used to analyze online course engagement data and predict whether a student is highly engaged or not. By applying techniques such as data preprocessing, visualization, and the Random Forest Classifier, the system can identify important patterns in student activity. This helps institutions and instructors make data-driven decisions to improve learning outcomes.

Import Libraries

```python
import warnings
warnings.filterwarnings('ignore')
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler,LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LinearRegression
from sklearn.metrics import accuracy_score,r2_score,mean_squared_error,classification_report,confusion_matrix
```

Import Dataset

```python
df=pd.read_csv('d:\DataScience\PROJECT\M L project\learn ML\online_course_engagement_data.csv')
```

Find Null values

```python
df.isna().sum()
```

Split Data

```python
x=df.drop(columns=['CourseCompletion'])
y=df['CourseCompletion']
```

Convert Categorical values into numerical

```python
Label_encoder=LabelEncoder()
for col in x.select_dtypes(include=['object']).columns:
    x[col]=Label_encoder.fit_transform(x[col])
```

Split into Train And Test Sets

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

Train model usiing Random Forest Classifier

```python
model=RandomForestClassifier(random_state=42)
model.fit(x_train,y_train)
```

Evaluate the Accuracy

```python
y_pred=model.predict(x_test)
accuracy=accuracy_score(y_test,y_pred)
print(f"Random Forest Model Accuracy:{accuracy*100:.2f}%\n")
```
